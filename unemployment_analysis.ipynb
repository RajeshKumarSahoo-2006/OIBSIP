```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option('display.max_columns', None)
```

## Load Dataset


```python
df = pd.read_csv("Unemployment in India.csv")

df.head()
```

## Dataset Information

```python
print("Shape of Dataset:", df.shape)

df.info()
```

```python
df.describe()
```

## Check Missing Values

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)

print("Dataset Shape After Cleaning:")
print(df.shape)
```

## Column Names

```python
print(df.columns)
```

## State Wise Unemployment Rate Analysis

```python
plt.figure(figsize=(14,6))

sns.barplot(
    x="Region",
    y=" Estimated Unemployment Rate (%)",
    data=df
)

plt.xticks(rotation=90)
plt.title("State Wise Unemployment Rate")
plt.tight_layout()
plt.show()
```

## Top 10 States with Highest Unemployment Rate

```python
state_unemployment = df.groupby("Region")[" Estimated Unemployment Rate (%)"].mean()

top_states = state_unemployment.sort_values(ascending=False).head(10)

plt.figure(figsize=(12,6))

sns.barplot(
    x=top_states.values,
    y=top_states.index
)

plt.title("Top 10 States with Highest Unemployment Rate")

plt.show()
```

## Monthly Unemployment Trend

```python
plt.figure(figsize=(14,6))

sns.lineplot(
    x=" Date",
    y=" Estimated Unemployment Rate (%)",
    data=df
)

plt.xticks(rotation=45)

plt.title("Monthly Unemployment Trend")

plt.tight_layout()

plt.show()
```

## Employment vs Unemployment Analysis

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    x=" Estimated Employed",
    y=" Estimated Unemployment Rate (%)",
    data=df
)

plt.title("Employment vs Unemployment")

plt.show()
```

## Distribution of Unemployment Rate

```python
plt.figure(figsize=(10,6))

sns.histplot(
    df[" Estimated Unemployment Rate (%)"],
    bins=20,
    kde=True
)

plt.title("Distribution of Unemployment Rate")

plt.show()
```

## Key Insights

### Insights

- Different states show different unemployment patterns.
- Some states have significantly higher unemployment rates.
- Employment levels vary across regions.
